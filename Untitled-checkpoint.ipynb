```python
df.drop_duplicates(inplace=True)
```

```python
# replacing the null and N|A values with blank

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
#removing extra spaces

df = df.apply(lambda x: x.str.strip() if x.dtype == 'str' else x) 
```

```python
# converting text into proper case

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# checking valid email adress

df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Convert Date Columns

df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce', dayfirst=True)
```

```python
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# removing records containing a negative quantity

df = df[df['Qty']>0]
```

```python
# Step 13: Fill Missing Values

df['Discount'] = df['Discount'].fillna (0)

df['Phone'] = df['Phone'].fillna('Unknown')

df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# Feature Engineering 
# Total Sales

df['Sales']=df['Qty']*df['Unit_Price']
```

```python
# converting the column first

df['Sales'] = pd.to_numeric(df['Sales'].astype(str).str.replace('$', '', regex=False), errors='coerce')
df['Discount'] = pd.to_numeric(df['Discount'].astype(str).str.replace('%', '', regex=False), errors='coerce')

# Calculate Net Amount
df['Net_Amount'] = df['Sales'] - (df['Sales'] * df['Discount'] / 100)
```

```python
# Profit (Example: 20%)

df['Profit']=df['Net_Amount']*0.20
```

```python
# Order Month

df['Month']=df['Order_Date'].dt.month_name()
```

```python
# Order Year

df["Year"]=df['Order_Date'].dt.year
```

```python
df
```

```python
# Weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```

```python
# exporting clean data into csv format

df.to_csv('Clean_Ecommerce.csv',index=False)
```

```python
from sqlalchemy import create_engine

engine = create_engine('postgresql://postgres:@localhost:/')

df.to_sql(
    name='ecommerce', 
    con=engine,
    if_exists='replace',
    index=False
)

print("Data imported successfully!")
```
